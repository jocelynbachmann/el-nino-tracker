```python
# Connect the data
import boto3, s3fs

DATA_FOLDER = "elnino"
PROFILE = "hack-team"
BUCKET  = "ocean-hackathon-2026"
fs = s3fs.S3FileSystem(profile=PROFILE)

who = boto3.Session(profile_name=PROFILE).client("sts").get_caller_identity()["Arn"]
TEAM = who.split("/")[-1]                    
DATA_DIR = f"{BUCKET}/data/elnino"          # your team's raw data (read-only)
TEAM_DIR = f"{BUCKET}/teams/team-elnino"    # your team's workspace (save results here)
```

```python
STICHED_DIR  = f"{TEAM_DIR}/processed"

for info in fs.ls(STICHED_DIR, detail=True):
    path = info["name"]
    name = path.rstrip("/").split("/")[-1]
    if info["type"] == "directory":
        print(f'  [folder]   {path}/')
    elif name:
        print(f'{info["size"]/1e9:8.2f} GB   {name}')
```
